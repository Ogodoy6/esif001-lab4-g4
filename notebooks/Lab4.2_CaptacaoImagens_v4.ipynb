## 1. Captura pela webcam (Google Colab)
Execute a célula e clique em **Capturar** quando estiver pronto. O navegador solicitará permissão para usar a câmera.

```python
from IPython.display import display, Javascript
from google.colab.output import eval_js
from base64 import b64decode
from PIL import Image
from io import BytesIO
import time
CAPTURAS=PASTA/'imagens'/'originais'
CAPTURAS.mkdir(parents=True,exist_ok=True)
def capturar(nome='grupo'):
    js = Javascript('''
    async function snap() {
      const stream = await navigator.mediaDevices.getUserMedia({video:true});
      const video = document.createElement('video'); video.srcObject=stream;
      await video.play();
      const button=document.createElement('button'); button.textContent='Capturar foto';
      document.body.appendChild(video); document.body.appendChild(button);
      await new Promise(resolve=>button.onclick=resolve);
      const canvas=document.createElement('canvas');
      canvas.width=video.videoWidth;canvas.height=video.videoHeight;
      canvas.getContext('2d').drawImage(video,0,0);
      stream.getTracks().forEach(t=>t.stop());video.remove();button.remove();
      return canvas.toDataURL('image/png');
    }''')
    display(js)
    url=eval_js('snap()')
    imagem=Image.open(BytesIO(b64decode(url.split(',')[1]))).convert('RGB')
    destino=CAPTURAS/f'{nome}_{int(time.time())}.png'
    imagem.save(destino)
    display(imagem)
    print('Salvo:',destino)
    return destino
```
